# Support Vector Machines

Support Vector Machines (SVM) is a supervised machine learning algorithm which can be used for both classification or regression challenges. However, it is mostly used in classification problems.

In this algorithm, we plot each data item as a point in n-dimensional space (where *n* is the number of features you have) with the value of each feature being the value of a particular coordinate. Then, we perform classification by finding the hyperplane that differentiates the two classes very well.

## Types of SVM Kernels:
- **Linear Kernel**
- **Polynomial Kernel**
- **Radial Basis Function Kernel (RBF)**
