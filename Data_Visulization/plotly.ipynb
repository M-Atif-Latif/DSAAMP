# Plotly

## Data Visualziation
Data Visualization is the graphical representation of information and data. By using visual elements like charts, graphs, and maps, data visualization tools provide an accessible way to see and understand trends, outliers, and patterns in data

## Plotly
- Plotly is a Python library which is used to design graphs, especially interactive graphs.

- It can plot various graphs and charts like histogram, barplot, boxplot, spreadplot, and many more.
- It is mainly used in data analysis as well as financial analysis. Plotly is an interactive visualization library.

### Benefits of using plolty:
- Plotly is interactive, the graphs can be zoomed in and out, downloaded as a PNG file, hovered over to see the data points, and much more.
- Plotly is compatible with Pandas DataFrames, making it easy to plot data directly from CSVs.
- Plotly can help making animated plots, which can be very useful for visualizing data over time.
- Plotly can be used to style interactive graphs with Jupyter notebook.
- Plotly with dash is a great choice for creating interactive dashboards that can be deployed on the web.

```python
# import library
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
import plotly.graph_objects as go
```

```python
# remove warning
import warnings
warnings.filterwarnings('ignore')
```

```python
# load dataset of dimond from seaborn
df = sns.load_dataset('diamonds')
df= df.sample(frac=0.1, random_state=10) # sample 10% of the data for faster processing
df.head()
```

```python
# reindex the data
df.reset_index(drop=True, inplace=True)
```

```python
df.head()
```

```python
# scatter plot
fig = px.scatter(df, x='carat', y='price', color='cut')
fig.update_traces(marker=dict(line=dict(width=1, color='DarkSlateGrey')))
fig.update_layout(title='Diamonds Price vs Carat', xaxis_title='Carat', yaxis_title='Price')
fig.show()
```

- we used color argument in it.

```python
# Plot a line chart (example: mean price by carat)
mean_price_by_carat = df.groupby('carat')['price'].mean().reset_index()
fig = px.line(mean_price_by_carat, x='carat', y='price', title='Mean Price by Carat')
fig.update_traces(line=dict(width=2, color='blue'))
fig.update_layout(title='Mean Price by Carat', xaxis_title='Carat', yaxis_title='Mean Price')
fig.show()
```

```python
# line plot
fig = px.line(df, x='carat', y='price', color='cut', line_group='cut', hover_name='clarity')
fig.update_traces(marker=dict(line=dict(width=1, color='DarkSlateGrey')))
fig.update_layout(title='Diamonds Price vs Carat', xaxis_title='Carat', yaxis_title='Price')
fig.show()
```

```python
# bar plot
fig = px.bar(df, x='cut', y='price', color='cut', title='Average Price by Cut')
fig.update_traces(marker=dict(line=dict(width=1, color='DarkSlateGrey')))
fig.update_layout(title='Average Price by Cut', xaxis_title='Cut', yaxis_title='Average Price')
fig.show()
```

```python
df_stats = df.groupby('clarity')['price'].agg(['mean', 'std']).reset_index()
df_stats.head()

# Bar plot for average price per cut category
fig = px.bar(df_stats, x='clarity', y='mean', error_y='std', 
             title='Average Price by Cut', color='clarity')
fig.show()
```

```python
# histogram plot
fig = px.histogram(df, x='price', color='cut', title='Carat Distribution by Cut')
fig.update_traces(marker=dict(line=dict(width=1, color='DarkSlateGrey')))
fig.update_layout(title='Carat Distribution by Cut', xaxis_title='Carat', yaxis_title='Count')
fig.show()
```

```python
# Density Contour plot
fig = px.density_contour(df, x='carat', y='price', color='cut', title='Density Contour of Carat vs Price')
fig.update_traces(contours_coloring='fill', contours_showlabels=True)
fig.update_layout(title='Density Contour of Carat vs Price', xaxis_title='Carat', yaxis_title='Price')
fig.show()
```

```python
# Box plot
fig = px.box(df, x='cut', y='price', color='cut', title='Box Plot of Price by Cut')
fig.update_traces(marker=dict(line=dict(width=1, color='DarkSlateGrey')))
# outlier points
points = 'suspected outliers'
fig.update_layout(title='Box Plot of Price by Cut', xaxis_title='Cut', yaxis_title='Price')
fig.show()
```

```python
# box plot
fig = px.box(df, x='cut', y='price', color='clarity',
             # outliers
            points='suspectedoutliers', # 'outliers', 'suspectedoutliers'
            title='Price Distribution by Cut',
            labels={'price':'Price ($)', 'cut':'Cut', 'clarity':'Clarity'}
            )
fig.show()
```

```python
# volin plot
fig = px.violin(df, x='cut', y='price', color='cut', box=True,
                 title='Violin Plot of Price by Cut')
fig.update_traces(marker=dict(line=dict(width=1, color='DarkSlateGrey')))
fig.show()
```

```python
# density plot
fig = px.density_contour(df, x='carat', y='price', color='cut',
                          title='Density Contour of Carat vs Price')
fig.update_traces(contours_coloring='fill', contours_showlabels=True)
fig.update_layout(title='Density Contour of Carat vs Price', xaxis_title='Carat', yaxis_title='Price')
fig.show()
```

```python
from plotly.subplots import make_subplots

# subplot
fig = make_subplots(rows=2, cols=2,
                    specs=[[{'type':'scatter'}, {'type':'scatter'}],
                           [{'type':'scatter'}, {'type':'scatter'}]],
                    subplot_titles=('Scatter Plot', 'Line Plot', 'Bar Plot', 'Box Plot'))
fig.add_trace(go.Scatter(x=df['carat'], y=df['price'], mode='markers', name='Scatter Plot'), row=1, col=1)
fig.add_trace(go.Scatter(x=df['carat'], y=df['price'], mode='lines', name='Line Plot'), row=1, col=2)
fig.add_trace(go.Bar(x=df['cut'], y=df['price'], name='Bar Plot'), row=2, col=1)
fig.add_trace(go.Box(x=df['cut'], y=df['price'], name='Box Plot'), row=2, col=2)
fig.update_layout(title='Subplots of Different Plot Types', height=800, width=800)
fig.show()
```

```python
from plotly.subplots import make_subplots

# Create subplots
fig = make_subplots(rows=1, cols=2, 
                    subplot_titles=('Carat vs Price', 
                                    'Density Heatmap of Carat vs Price'))

# Scatter plot of carat vs price
scatter = px.scatter(df, x='carat', y='price').data[0]
fig.add_trace(scatter, row=1, col=1)

# Density heatmap of carat vs price
heatmap = px.density_heatmap(df, x='carat', y='price').data[0]
fig.add_trace(heatmap, row=1, col=2)

# Update layout
fig.update_layout(title_text='Carat vs Price and Density Heatmap of Carat vs Price')
fig.show()
```

```python
#heatmap plot
fig = px.density_heatmap(df, x='carat', y='price', z='cut', title='Density Heatmap of Carat vs Price')
fig.update_layout(title='Density Heatmap of Carat vs Price', xaxis_title='Carat', yaxis_title='Price')
fig.show()
```

```python
# creat heatmap of correlation matrix of only numerical columns
corr = df[['carat', 'depth', 'table', 'price', 'x', 'y', 'z']].corr()
fig = go.Figure(data=go.Heatmap(x=corr.index.values, 
                                y=corr.columns.values, 
                                z=corr.values, 
                                colorscale='Reds')) 
# colorscale can be 'Viridis', 'Cividis', 'Blues', 'Greens', 'Reds', 'Oranges', 'YlOrRd', 'YlGnBu', 'RdBu', 'Picnic', 'Rainbow', 'Portland', 'Jet', 'Hot', 'Blackbody', 'Earth', 'Electric', 'Plasma', 'Magma', 'Inferno', 'Cividis'
fig.show()
```

### Save Animation plots

```python
import plotly.express as px
import seaborn as sns
import os
from moviepy.editor import ImageSequenceClip

# Load the diamonds dataset
df = sns.load_dataset('diamonds')

# Check for missing values
if df.isnull().any().any():
    raise ValueError("The dataset contains missing values. Please handle them before plotting.")

# Create a directory to save frames
frames_dir = "frames"
os.makedirs(frames_dir, exist_ok=True)

# Extract unique values for the 'cut' column to iterate over
unique_cuts = df['cut'].unique()

# Save each frame as an image
for i, cut in enumerate(unique_cuts):
    # Filter the data for the current cut
    filtered_df = df[df['cut'] == cut]
    
    # Ensure filtered data is valid
    if filtered_df.empty:
        continue

    # Create the scatter plot for the current cut
    frame_fig = px.scatter(
        filtered_df, x='carat', y='price',
        color='color',
        title=f'Carat vs Price for {cut} Cut',
        size='depth',
        hover_data=['clarity']
    )
    
    # Save the frame image using Kaleido
    frame_filename = os.path.join(frames_dir, f"frame_{i:03d}.png")
    frame_fig.write_image(frame_filename, engine="kaleido")

# Create a GIF or MP4 using moviepy
image_files = [os.path.join(frames_dir, f) for f in sorted(os.listdir(frames_dir)) if f.endswith('.png')]
clip = ImageSequenceClip(image_files, fps=2)  # Adjust fps as needed
clip.write_gif("./03_plotly/animated_plot.gif", fps=2)    # Save as GIF
clip.write_videofile("./03_plotly/animated_plot.mp4", fps=1)  # Save as MP4
# save HD video
clip.write_videofile("./03_plotly/animated_plot_HD.mp4", fps=1, codec="libx264", preset="ultrafast", bitrate="3000k")
# Display the GIF in Jupyter Notebook
from IPython.display import Image
Image("./03_plotly/animated_plot.gif")
```
