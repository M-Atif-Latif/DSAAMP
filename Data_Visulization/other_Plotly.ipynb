```python
# import the libraries
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import plotly.express as px
import plotly.graph_objects as go
import plotly.io as pio
import numpy as np
```

```python
# Load the diamonds dataset
df = sns.load_dataset('diamonds')
```

```python
df.head()
```

```python
# scatter plot
fig = px.scatter(df, x='carat', y='price', color='cut', size='depth', hover_name='cut')
fig.update_traces(marker=dict(line=dict(width=1, color='DarkSlateGrey')))
fig.update_layout(title='Diamonds Dataset Scatter Plot', xaxis_title='Carat', yaxis_title='Price')
# taking log
df['log_price'] = np.log(df['price'])
df['log_carat'] = np.log(df['carat'])
fig.show()
```

```python
from plotly.subplots import make_subplots

# subplot
fig = make_subplots(rows=1, cols=2, specs=[[{'type': 'scatter'}, {'type': 'scatter'}]])
fig.add_trace(go.Scatter(x=df['carat'], y=df['price'], mode='markers', name='Carat vs Price'), row=1, col=1)
fig.add_trace(go.Scatter(x=df['log_carat'], y=df['log_price'], mode='markers', name='Log Carat vs Log Price'), row=1, col=2)
fig.update_layout(title_text='Diamonds Dataset Scatter Plot', title_x=0.5)
fig.update_xaxes(title_text='Carat', row=1, col=1)
fig.update_yaxes(title_text='Price', row=1, col=1)
fig.update_xaxes(title_text='Log Carat', row=1, col=2)
fig.update_yaxes(title_text='Log Price', row=1, col=2)
fig.show()
```

```python
# scatter plot with dual y-axes
fig = make_subplots(specs=[[{'secondary_y': True}]])
fig.add_trace(go.Scatter(x=df['carat'], y=df['price'], mode='markers', name='Carat vs Price'), secondary_y=False)
fig.add_trace(go.Scatter(x=df['carat'], y=df['log_price'], mode='markers', name='Carat vs Log Price'), secondary_y=True)
fig.update_layout(title_text='Diamonds Dataset Scatter Plot with Dual Y-Axes', title_x=0.5)
fig.update_xaxes(title_text='Carat')
fig.update_yaxes(title_text='Price', secondary_y=False)
fig.update_yaxes(title_text='Log Price', secondary_y=True)
fig.show()
```

```python
# Bullet chart for price distribution
fig = go.Figure(go.Indicator(
    mode="gauge+number",
    value=df['price'].mean(),
    title={"text": "Average Diamond Price"},
    gauge={"axis": {"range": [0, max(df['price'])]}}
))
fig.show()
```
