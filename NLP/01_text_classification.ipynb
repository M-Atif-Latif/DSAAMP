```python
# This script demonstrates how to perform sentiment analysis using a pre-trained model from Hugging Face's Transformers library.
# The model used is DistilBERT fine-tuned on the SST-2 dataset.
from transformers import pipeline

# one-liner sentiment analysis on the SST-2-finetuned DistilBERT
classifier = pipeline(
    task="text-classification",
    model="distilbert-base-uncased-finetuned-sst-2-english"
)

texts = [
    "I absolutely loved this movie!",
    "This was a waste of two hours."
]
print(classifier(texts))
```

## Read and classify Public (Dummy) comments using a pre-trained model

```python
from transformers import pipeline
# one-liner sentiment analysis on the SST-2-finetuned DistilBERT
classifier = pipeline(
    task="text-classification",
    model="distilbert-base-uncased-finetuned-sst-2-english"
)

# read csv file of comments and run thr classifier
import pandas as pd
df = pd.read_csv("youtube_comments_dummy.csv")
texts = df["comment"].tolist()
results = classifier(texts)
# add results to the dataframe
df["sentiment"] = [result['label'] for result in results]
# save the dataframe to a new csv file
df.to_csv("comments_with_sentiment.csv", index=False)
df.head()

# create a pie chart
import matplotlib.pyplot as plt
# count the number of positive and negative comments
sentiment_counts = df["sentiment"].value_counts()
# show the sentiment
# create a pie chart
plt.figure(figsize=(8, 6))
sentiment_counts.plot(kind='pie', autopct='%1.1f%%', startangle=90)
plt.title("Sentiment Analysis of Comments")
plt.ylabel("")  # Hide the y-label
plt.show()
```
